```python
import os
import numpy as np
import pandas as pd

# Path to the keyframes directory
keyframes_dir = "keyframes"

counts = []

for folder in sorted(os.listdir(keyframes_dir)):
    folder_path = os.path.join(keyframes_dir, folder)

    if not os.path.isdir(folder_path):
        continue

    # Count image files (modify extensions if needed)
    n_frames = len([
        f for f in os.listdir(folder_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp"))
    ])

    counts.append({
        "Video": folder,
        "Keyframes": n_frames
    })

df = pd.DataFrame(counts)

print("="*60)
print(f"Total videos           : {len(df)}")
print(f"Total keyframes        : {df['Keyframes'].sum()}")
print(f"Minimum keyframes      : {df['Keyframes'].min()}")
print(f"Maximum keyframes      : {df['Keyframes'].max()}")
print(f"Mean                   : {df['Keyframes'].mean():.2f}")
print(f"Median                 : {df['Keyframes'].median():.2f}")
print(f"Standard deviation     : {df['Keyframes'].std():.2f}")
print(f"Variance               : {df['Keyframes'].var():.2f}")
print("="*60)

print("\nPercentiles")
for p in [50, 75, 90, 95, 97, 99]:
    print(f"{p:2d}th percentile : {np.percentile(df['Keyframes'], p):.2f}")

print("\nFive-number summary")
print(df["Keyframes"].describe())

print("\nFrequency distribution")
distribution = df["Keyframes"].value_counts().sort_index()

print("Keyframes : Number of Videos")
print("-" * 30)

for keyframes, num_videos in distribution.items():
    print(f"{keyframes:2d} : {num_videos}")

# Save statistics
df.to_csv("keyframe_statistics.csv", index=False)

print("\nStatistics saved to keyframe_statistics.csv")
```

```python
import os
import numpy as np
import tqdm
from PIL import Image

def load_and_preprocess_images(numofframes,
                               train_folders,
                               train_labels,
                               img_size=(224, 224),
                               data_dir="keyframes"):

    train_data = []
    train_labels_final = []
    new_height, new_width = img_size

    for j, folder in enumerate(tqdm.tqdm(train_folders)):

        temp_data = []
        folder_path = os.path.join(data_dir, str(folder))

        # Get all jpg files and sort numerically
        frame_files = sorted(
            [f for f in os.listdir(folder_path) if f.lower().endswith(".jpg")],
            key=lambda x: int(os.path.splitext(x)[0])
        )

        if len(frame_files) == 0:
            raise ValueError(f"No frames found in {folder_path}")

        # -------------------------------------------------
        # If fewer frames than required, replicate evenly
        # -------------------------------------------------
        if len(frame_files) < numofframes:
            indices = np.linspace(0, len(frame_files) - 1, numofframes)
            indices = np.round(indices).astype(int)
            selected_frames = [frame_files[i] for i in indices]
        else:
            # Uniformly sample the required number of frames
            indices = np.linspace(0, len(frame_files) - 1, numofframes, dtype=int)
            selected_frames = [frame_files[i] for i in indices]

        # Load selected frames
        for frame_name in selected_frames:
            image_path = os.path.join(folder_path, frame_name)

            image = Image.open(image_path)
            image = image.resize((new_width, new_height))
            image = np.array(image)

            temp_data.append(image)

        train_data.append(temp_data)
        train_labels_final.append(train_labels[j])

    return np.array(train_data), np.array(train_labels_final)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
from tensorflow.keras.optimizers import Adam
def build_efficientnet_gru_model(num_frames, input_shape=(224, 224, 3), num_classes=101,
                                 gru_units_1=128, gru_units_2=64, dropout_rate=0.5):
    """
    Builds an EfficientNet + Bidirectional GRU model for video classification.
    
    Args:
        num_frames (int): Number of frames in the video sequence.
        input_shape (tuple): Shape of each video frame (height, width, channels).
        num_classes (int): Number of output classes.
        gru_units_1 (int): Number of units in the first GRU layer.
        gru_units_2 (int): Number of units in the second GRU layer.
        dropout_rate (float): Dropout rate for regularization.
    
    Returns:
        keras.Model: Compiled Keras model.
    """

    # Load pre-trained EfficientNetB3 model (excluding classification layers)
    base_model = keras.models.load_model("/workspace/data/data/UCFcodes/3rdpaper/Efficient_.h5")  # Load EfficientNetB3 from .h5 file
    base_model = models.Model(inputs=base_model.input, outputs=base_model.layers[-2].output)  # Remove last layer
    base_model.trainable = False  # Freeze EfficientNetB3 layers (can be set to True for fine-tuning)

    # Define model input shape: (batch_size, num_frames, height, width, channels)
    video_input = keras.Input(shape=(num_frames, *input_shape), name="video_input")

    # Apply EfficientNetB3 to each frame separately using TimeDistributed
    reshaped_frames = layers.TimeDistributed(base_model)(video_input)
    reshaped_frames = layers.TimeDistributed(layers.Flatten())(reshaped_frames)  # Flatten the extracted features

    # GRU layers for temporal modeling
    gru_output = layers.Bidirectional(layers.GRU(units=gru_units_1, return_sequences=True))(reshaped_frames)
    gru_output = layers.BatchNormalization()(gru_output)
    gru_output = layers.Dropout(dropout_rate)(gru_output)

    gru_output = layers.Bidirectional(layers.GRU(units=gru_units_2))(gru_output)
    gru_output = layers.BatchNormalization()(gru_output)
    gru_output = layers.Dropout(dropout_rate)(gru_output)

    # Classification layer
    output = layers.Dense(units=num_classes, activation="softmax")(gru_output)

    # Define the model
    model = keras.Model(inputs=video_input, outputs=output, name=f"EfficientNetB3_GRU_{num_frames}Frames")

    # Compile the model
    model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])

    return model

# Example usage:
num_frames = 7  # Can be changed based on the dataset
model = build_efficientnet_gru_model(num_frames)
model.summary()
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models

def build_efficientnet_gru_model(input_shape=(224, 224, 3),
                                 num_classes=101,
                                 gru_units_1=128,
                                 gru_units_2=64,
                                 dropout_rate=0.5):

    # Load pretrained EfficientNet feature extractor
    base_model = keras.models.load_model(
        "/workspace/data/data/UCFcodes/3rdpaper/Efficient_.h5"
    )

    base_model = models.Model(
        inputs=base_model.input,
        outputs=base_model.layers[-2].output
    )

    base_model.trainable = False

    video_input = keras.Input(
    shape=(None, *input_shape),
    name="video_input",
    ragged=True
    )
    # Feature extraction
    x = layers.TimeDistributed(base_model)(video_input)

    x = layers.TimeDistributed(layers.Flatten())(x)

    # Ignore padded frames (assumes padding is all zeros)
    x = layers.Masking(mask_value=0.0)(x)

    # Temporal modelling
    x = layers.Bidirectional(
        layers.GRU(
            gru_units_1,
            return_sequences=True
        )
    )(x)

    x = layers.BatchNormalization()(x)
    x = layers.Dropout(dropout_rate)(x)

    x = layers.Bidirectional(
        layers.GRU(
            gru_units_2,
            return_sequences=False
        )
    )(x)

    x = layers.BatchNormalization()(x)
    x = layers.Dropout(dropout_rate)(x)

    outputs = layers.Dense(
        num_classes,
        activation="softmax"
    )(x)

    model = keras.Model(video_input, outputs)

    model.compile(
        optimizer="adam",
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


model = build_efficientnet_gru_model()

model.summary()
```
